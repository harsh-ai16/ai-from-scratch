```
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4,5,6,7,8])
y = np.array([1,2,3,4,5,6,7,8])

coeffs = np.polyfit(x, y, 1)
xline = np.linspace(1, 8, 100)
yline = coeffs[0] * xline + coeffs[1]

plt.figure(figsize=(6,8))

plt.scatter(
    x, y,
    color="blue",
    s=80,
    alpha=0.9,
    label="Data Points"
)

plt.plot(
    xline, yline,
    color="green",
    linewidth=1.5,
    label="Model Line"
)

plt.xlabel("X values", fontsize=12)
plt.ylabel("Y values", fontsize=12)
plt.title("Scatter Plot with Linear Model", fontsize=14)

plt.legend()
plt.grid(True, alpha=0.5)

plt.show()
```

```
maths=[56,34,89,99,78,80]
science=[89,67,90,78,72,93]

plt.style.use('_mpl-gallery')
fig, ax = plt.subplots()

ax.bar(science,maths, width=1, edgecolor="white", linewidth=0.7)

ax.set(title="Bar Plot")
ax.set_xlabel("science")
ax.set_ylabel("maths")
ax.grid(True,alpha=0.7)

plt.show()
```
