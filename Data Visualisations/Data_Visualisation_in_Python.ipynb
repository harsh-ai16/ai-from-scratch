```python
import numpy as np
import matplotlib.pyplot as plt

temperature = np.array([20, 22, 24, 26, 28, 30, 32, 34])
sales = np.array([80, 85, 88, 95, 100, 105, 110, 120])


plt.scatter(temperature, sales,label="Sales")
plt.xlabel("Temperature (°C)")
plt.ylabel("Ice Cream Sales")
plt.title("Temperature vs Ice Cream Sales")
m,c=2.86,22.26
line=m*temperature+c
plt.plot(temperature, line, color="green", label="Our Hint")
plt.legend()
plt.show()

# Add a line equation y=m*x+c to the below plot
```

```python
budgets = [20, 35, 50, 65, 80, 100, 120, 150]
revenues = [45, 60, 90, 110, 140, 180, 210, 260]
coeffs=np.polyfit(budgets,revenues,1)
x_line=np.linspace(20,150,100)
y_line=coeffs[0]*x_line+coeffs[1]
plt.figure(figsize=(8,6))
plt.scatter(budgets,revenues,color="blue",s=80,alpha=0.7,label="Data Points")
plt.plot(x_line,y_line, color="green", linewidth=2, label="Predicted value")
plt.xlabel("Budget")
plt.ylabel("Revenues")
plt.title("Budget vs Revenues")
plt.legend()
plt.grid(True,alpha=0.4)
plt.show()
```
