```python
import matplotlib.pyplot as plt
import numpy as np
xdata=np.linspace(-10,10,100)
yconvex= (xdata**3)-3*xdata
# ynonconvex=abs(xdata)
y_nonconvex=np.sin(xdata)

fig,ax=plt.subplots(1,2, figsize=(8,10))

ax[0].plot(xdata,yconvex,color="red")
ax[0].grid(True)
# ax[1].plot(xdata,ynonconvex,color="blue")
# ax[1].grid(True)
ax[1].plot(xdata,y_nonconvex,color="green")
ax[1].grid(True)
plt.show()
```
