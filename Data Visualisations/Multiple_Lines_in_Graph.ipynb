```python
from matplotlib.lines import lineStyles
import matplotlib.pyplot as plt
import numpy as np
import random

x=np.linspace(0,10,50)
y_scatter=2*x+np.random.normal(0,2,50)
line1=2*x+1
line2=1.5*x+3

plt.figure(figsize=(10,8))
plt.scatter(x,y_scatter,color="red",s=80,alpha=0.8,label="Data Points")
plt.plot(x,line1,color="blue",linestyle="--",linewidth=3,label="Trend line 1")
plt.plot(x,line2,color="green",linewidth=3,label="Trend line 2")
plt.xlabel("X value",fontsize=12)
plt.ylabel("Y value",fontsize=12)
plt.title("Multiple lines showing trends with scatter plot",fontsize=18,fontweight="bold")
plt.legend(frameon=True , shadow=True)
plt.grid(True,alpha=0.4)
plt.show()
```
