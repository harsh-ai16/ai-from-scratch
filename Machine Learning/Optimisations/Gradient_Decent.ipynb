```python
import matplotlib.pyplot as plt
import numpy as np

def f(x):
  return (x-3)**2+1

def df_dx(x):
  return 2 * (x-3)

# Gradient Decent
x_start=-0.5
learning_rate=0.1
iterations=25

x=x_start
x_history=[x]
y_history=[f(x)]
for i in range(iterations):
  gradient=df_dx(x)
  x=x-learning_rate*gradient

  print(f"iterations {i+1},  x: {x:.3f},  f(x): {f(x):.3f},  Gradient: {gradient:.3f}")
  x_history.append(x)
  y_history.append(f(x))

xplot=np.linspace(-1,6,100)
yplot=f(xplot)

plt.plot(xplot,yplot,label="(x-3)**2+1")
plt.plot(x_history,y_history,'ro',label=" Gradient Decent LR=0.1")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True,alpha=0.4)
plt.legend()
plt.show()
```
